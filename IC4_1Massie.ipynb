```python
%%sql
CREATE TABLE products (
  product_id INTEGER PRIMARY KEY,
  name TEXT,
  price REAL
) STRICT;
```

```python
%%sql
INSERT INTO products (name, price)
VALUE ('Superbowl Jersey', 120.00);
```

```python
%%sql
SELECT * FROM products WHERE price > 50;
```

```python
%%sql
UPDATE products
SET price = 99.99
WHERE name = 'Superbowl Jersey';
```
